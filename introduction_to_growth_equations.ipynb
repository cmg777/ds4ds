# Introduction to Growth Equations

# Overview


## 💡 Basic Macroeconomic Growth Concepts
- Growth theory analyzes long-run output trends.
- Models use variables like GDP, capital, labor, and technology.
- Equations help explain and predict economic trajectories.

---

## 🧮 Output and the Production Function
- Output ($Y$) is a function of capital ($K$), labor ($L$), and technology ($A$):

  $$
  Y = f(K, L, A)
  $$

---

## 🏗️ Capital Accumulation Equation
- Investment drives capital stock growth:

  $$
  K_{t+1} = (1 - \delta)K_t + I_t
  $$

- $ \delta $: depreciation rate  
- $ I_t $: gross investment

---

## 👨‍💼 Labor Force Dynamics
- Labor grows with population and participation:

  $$
  L_{t+1} = (1 + n)L_t
  $$

- $ n $: labor force growth rate

---

## 💾 Savings Rate and Investment
- A portion of output ($Y_t$) is saved and invested:

  $$
  I_t = sY_t
  $$

- $ s $: savings rate

---

## 📉 Investment Efficiency (ICOR)
- Measures how effectively capital turns into output:

  $$
  ICOR = \frac{K_{t+1} - K_t}{Y_{t+1} - Y_t}
  $$

- Lower ICOR = higher investment efficiency

---

## 🧬 Technology in Growth Equations
- Technology ($A$) improves productivity over time:

  $$
  A_{t+1} = (1 + g)A_t
  $$

- $ g $: technological growth rate

---

## 📊 Balanced Growth Path Equation
- On a steady path, output grows with capital, labor, and tech:

$$
\frac{\dot{Y}}{Y}=\frac{\dot{K}}{K}=\alpha \frac{\dot{A}}{A}+(1-\alpha) \frac{\dot{L}}{L}
$$

- Based on $ Y = AK^\alpha L^{1 - \alpha} $

---

## 🔄 Solow Growth Model Insight
- In steady state, per-worker output grows via tech:

  $$
  \frac{\dot{y}}{y} = g
  $$

- $ y $: output per worker

---

## ✅ Conclusion
- Growth equations link $Y$ to $K$, $L$, $A$, and $s$.
- Capital, labor, savings, and tech are key growth drivers.
- Solow model emphasizes tech as the engine of long-run growth.


# Algorithmic representation

To compute growth under certain assumptions, algorithmic methods can be applied. Below is a simple loop-based model using constant savings and labor growth:


### **🔢 Inputs**
- Initial capital: $K_0$
- Initial labor: $L_0$
- Savings rate: $s$
- Depreciation rate: $\delta$
- Labor growth rate: $n$
- Time horizon: $T$



### **🧮 Output**
- Projected output: $Y_t$ over time



### **🔁 Pseudocode**

**While** time $t < T$ **do**  
  Compute investment: $I_t = sY_t$  
  Update capital: $K_{t+1} = (1 - \delta)K_t + I_t$  
  Update labor: $L_{t+1} = (1 + n)L_t$  
  Compute output: $Y_{t+1} = A_t K_{t+1}^\alpha L_{t+1}^{1 - \alpha}$  
  Advance time: $t = t + 1$



# Code

## Setup

```python
import numpy as np
import pandas as pd
import plotly.express as px
```

## Define functions

```python
def economic_output(K, L, A, alpha):
    """
    Calculate economic output using the Cobb-Douglas production function.
    :param K: Capital stock.
    :param L: Labor input.
    :param A: Technology level or Total Factor Productivity.
    :param alpha: Output elasticity of capital.
    :return: Total economic output.
    """
    return A * (K ** alpha) * (L ** (1 - alpha))
```

```python
def capital_accumulation(K_t, I_t, delta):
    """
    Calculate new capital stock after accounting for depreciation and investment.
    :param K_t: Existing capital at time t.
    :param I_t: Gross investment at time t.
    :param delta: Depreciation rate.
    :return: New capital stock for time t+1.
    """
    return (1 - delta) * K_t + I_t
```

```python
def labor_dynamics(L_t, n):
    """
    Calculate next period's labor force size with growth rate.
    :param L_t: Current labor force size.
    :param n: Labor force growth rate.
    :return: Labor force size at time t+1.
    """
    return (1 + n) * L_t
```

```python
def investment_from_savings(Y_t, s):
    """
    Compute investment as a function of savings rate and output.
    :param Y_t: Total economic output at time t.
    :param s: Savings rate.
    :return: Investment at time t.
    """
    return s * Y_t
```

```python
def technological_progress(A_t, g):
    """
    Determine technology level for next period given a growth rate.
    :param A_t: Current technology level.
    :param g: Growth rate of technology.
    :return: Technology level at time t+1.
    """
    return (1 + g) * A_t
```

## Example 1

```python
# Set parameter values
alpha = 0.3
delta = 0.1
n = 0.01
s = 0.2
g = 0.02
```

```python
# Set initial values
K_0 = 1000 # Initial Capital
L_0 = 100  # Initial Labor
A_0 = 1.0  # Initial Technology Level
```

```python
# Simulating over time
T = 10 # Number of time periods
K_t = K_0
L_t = L_0
A_t = A_0

for t in range(T):
    # Calculate economic variables for the current period
    Y_t = economic_output(K_t, L_t, A_t, alpha)
    I_t = investment_from_savings(Y_t, s)

    # Calculate next period's values
    L_t1 = labor_dynamics(L_t, n)
    A_t1 = technological_progress(A_t, g)
    K_t1 = capital_accumulation(K_t, I_t, delta)

    # Update values for the next period
    K_t = K_t1
    L_t = L_t1
    A_t = A_t1

    # Print results
    print(f"Period {t+1}:")
    print(f"Output (Y_{t+1}): {Y_t:.2f}")
    print(f"Capital (K_{t+1}): {K_t:.2f}")
    print(f"Labor (L_{t+1}): {L_t:.2f}")
    print(f"Technology (A_{t+1}): {A_t:.2f}")
    print("-" * 20)
```

## Example 2

```python
# --- Parameters and Initial Values ---
alpha = 0.3
delta = 0.1
n = 0.01
s = 0.2
g = 0.02
K_0 = 1000 # Initial Capital
L_0 = 100  # Initial Labor
A_0 = 1.0  # Initial Technology Level
T = 25 # Number of time periods for simulation
```

```python
# --- Simulation with History Tracking ---
K_t, L_t, A_t = K_0, L_0, A_0
```

```python
# Initialize lists to store the history of variables, including the initial state (t=0)
time_periods = list(range(T + 1)) # Represents time 0, 1, ..., T
K_hist = [K_t]
L_hist = [L_t]
A_hist = [A_t]
# Calculate initial output Y_0
Y_0 = economic_output(K_t, L_t, A_t, alpha)
Y_hist = [Y_0]
```

```python
# Run the simulation loop for T periods (generating results for t=1 to t=T)
for t in range(T):
    # 1. Calculate current period's output (needed for investment)
    Y_t = economic_output(K_t, L_t, A_t, alpha)
    I_t = investment_from_savings(Y_t, s)

    # 2. Calculate next period's state variables (K_t+1, L_t+1, A_t+1)
    K_t1 = capital_accumulation(K_t, I_t, delta)
    L_t1 = labor_dynamics(L_t, n)
    A_t1 = technological_progress(A_t, g)

    # 3. Update state for the next iteration
    K_t, L_t, A_t = K_t1, L_t1, A_t1

    # 4. Calculate Output for the *new* state (Y_t+1)
    Y_t1 = economic_output(K_t, L_t, A_t, alpha)

    # 5. Store results for the *next* period (corresponding to index t+1 in lists)
    K_hist.append(K_t)
    L_hist.append(L_t)
    A_hist.append(A_t)
    Y_hist.append(Y_t1) # Store the output corresponding to the new state
```

```python
# --- Create a Pandas DataFrame for easy plotting ---
df = pd.DataFrame({
    'Time': time_periods,
    'Capital (K)': K_hist,
    'Labor (L)': L_hist,
    'Technology (A)': A_hist,
    'Output (Y)': Y_hist
})
df
```

```python
# Plot Capital (K) over Time
fig_K = px.line(df, x='Time', y='Capital (K)',
                title='Capital Stock (K) Over Time',
                labels={'Capital (K)': 'Capital Stock'},
                markers=True) # Add markers to see individual points
fig_K.show()
```

```python
# Plot Labor (L) over Time
fig_L = px.line(df, x='Time', y='Labor (L)',
                title='Labor Force (L) Over Time',
                labels={'Labor (L)': 'Labor Force Size'},
                markers=True)
fig_L.show()
```

```python
# Plot Technology (A) over Time
fig_A = px.line(df, x='Time', y='Technology (A)',
                title='Technology Level (A) Over Time',
                labels={'Technology (A)': 'Technology Level (TFP)'},
                markers=True)
fig_A.show()
```

```python
# Plot Output (Y) over Time
fig_Y = px.line(df, x='Time', y='Output (Y)',
                title='Economic Output (Y) Over Time',
                labels={'Output (Y)': 'Total Output'},
                markers=True)
fig_Y.show()
```
